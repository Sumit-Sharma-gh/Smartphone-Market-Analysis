## Expensive Phone

```python
df.loc[df['price_inr'].idxmax(), ['phone_name', 'price_inr']]
```

## Cheapest Phone

```python
df.loc[df['price_inr'].idxmin(), ['phone_name', 'price_inr']]
```

## Powerful Battery(MAh)

```python
df.loc[df['battery'].idxmax(), ['phone_name', 'battery']]
```

## Recently Launched Phone

```python
recently_launched = df.loc[df['launch_date'].idxmin(), ['phone_name', 'launch_date']]
recently_launched
```

## Average Price of Each Brands

```python
df.groupby('brand')['price_inr'].mean().reset_index(name = 'Avg_Price')
```

## Count Of All Brands

```python
df['brand'].value_counts().reset_index(name = 'Total_Counts')
```

## Budget Phones Unnder 12000

```python
df[df['price_inr'] <= 12000].sort_values('price_inr')[['phone_name', 'price_inr']]
```

```python
df[df['price_inr'] >= 100000].sort_values('price_inr')[['phone_name', 'price_inr']]
```

## Average Price Of Each Display Type Phone

```python
df.groupby('display_type')['price_inr'].mean().reset_index(name = 'Average_price INR')
```

```python
import matplotlib.pyplot as plt

df['brand'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Brands by Number of Phones')
plt.xlabel('Brand')
plt.ylabel('Count')

plt.show()
```

```python
top10 = df.nlargest(10, 'price_inr')

plt.barh(top10['phone_name'], top10['price_inr'])

plt.title('Top 10 Expensive Phones')
plt.xlabel('Price')
plt.ylabel('Phone')

plt.show()
```
